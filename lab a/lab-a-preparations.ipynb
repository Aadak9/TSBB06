## Question 3: The DLT Constraints

### Review of the Derivation

In the lectures, we derived the Direct Linear Transformation (DLT) constraints for the homography estimation problem in the following way:

Given homogeneous point correspondences y_k ↔ y_k', the objective is to find a homography matrix H such that y_k' is proportional to H y_k.
Using the cross product to rewrite this last relation, we obtain

y_k' ~ H y_k  iff  y_k' × H y_k = 0  iff  [y_k']_× H y_k = 0.

Rewriting this using the vec operator and the Kronecker product yields the DLT constraint

(y_k^T ⊗ [y_k']_×) vec(H) = 0,

which is a system of three linear equations in the nine unknowns in H.
These three equations are always linearly dependent, and as a result, one of the equations can always be eliminated.
For this reason, it is common to only use the first two equations (this is safe if all y_k and y_k' are proper points).
